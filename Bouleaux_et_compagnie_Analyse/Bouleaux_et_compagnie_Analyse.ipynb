## Détermination des valeurs abérrantes

```python
#On calcul les bornes des valeurs aberrantes, leur proportion dans le dataframe complet
# et on crée un dataset sans les valeurs abérrantes
inf, sup, nva = abbvalues(df, "Circonference_cm", 'oui')
df_sva = df[df["Circonference_cm"] < sup]

inf, sup, nva = abbvalues(df, "Hauteur_m", 'oui')
df_sva = df_sva[df_sva["Hauteur_m"] < sup]
```
