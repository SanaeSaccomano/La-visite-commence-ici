## Extraction des features

La méthode SIFT permet d'extraire les points clés (les features) et les descripteurs de l'image. 
Le point-clé est un point d'intérêt de l'image : un bord ou un coin de formes.
Le descripteur est le vecteur qui décrit le voisinage de son point-clé, la zone d'intérêt.

Les points-clés correspondants peuvent être détectés avec les détecteurs Cany et Harris-Stephens qui se basent sur l'étude du gradient. L'idée est de chercher les zones où le module du gradient d'intensité varie brusquement.

La méthode SIFT (Scale-invariant feature transform) (transformation de caractéristiques visuelles invarianteà l'échelle en français), permet de s'affranchir des limitations liées au changement d'échelle de la localisation des coins avec le détecteur Harris-Stephens.

<span class="mark">Pour détecter un point-clé, l'algorithme sur le principe de la création de l'espace des échelles et de la création des DoG (Difference of Gaussians)</span>. Par comparaison entre un point et son voisinnage (dans le plan mais aussi sur les plans au dessus et au dessous de lui), l'algorithme trouve des extrêmes locaux qui deviennent des points-clés potentiels qui sont sélectionnés ou non suivants différents critères (points aux extrémités, contraste *etc* ..).

L'algorithme SIFT détecte des points-clés qui sont décrits par des descripteurs.

### Extraction des descripteurs de quelques images

Testons l'algorithme SIFT pour un ensemble de 3 photos. Nous demandons à l'algorithme de se limiter à 300 descripteurs pour chacune des photos. 

Le descripteur correspond au voisinage du point clé. C'est la zone d'intérêt qui sera comparée aux autres images pour établir des points communs suivant les labels.
Ce descripteur est caractérisé par son orientation. Il s'agit de la direction prédominante du gradient autour du point d'intérêt.  Ensuite, la construction du descripteur se fait de la façon suivante :
- Une fenêtre de taille 12σ×12σ  pixels (proportionnelle à son niveau de flou de déctection donc) centrée autour d'un point-clé est découpée en 16 cellules (de taille 3σ×3σ  pixels).
- Pour chaque cellule, l'algorithme construit un histogramme des orientations du gradient (chaque orientation du gradient est soustrait à l'orientation dominante). 
- Chaque histogramme compte 8 classes de 45° chacune. Pour rendre compte de la proximité du pixel avec le point clé, l'amplitude est multipliée par une fonction gaussienne 2D. 
- Chaque cellule a son histogramme de 8 valeurs, ce qui fait 16*8 = 128 valeurs pour un descripteur. Ces valeurs sont stockées dans le vecteur "descripteur" donné en sortie de la fonction detectAndCompute() de opencv.

```python
descripteurs = visu_keypoints(photoch_acad, nb_photo=3, nb_descripteurs=300)
```

# Réduction de dimensions

## L'analyse en composantes principales

L'analyse en composantes principales permet de réduire le nombre de variables qui décrivent une base de données. L'algorithme permet de déterminer le nombre de variables à garder pour le pourcentage de niveau de variance que l'on souhaite conserver. La PCA permet de construire de nouvelles variables (les composantes principales d'intertie, notées F) qui sont des combinaisons linéaires des variables initiales. Ces nouvelles variables sont le moins corrélées possible entre elles. 

```python
features_pca = pca_descripteurs(features_images)
```

Cette méthode va permettre une meilleure séparation des données via le t-SNE et en diminue le temps de calcul.

## Réduction de dimension par t-SNE

La réduction de dimension en 2 composantes t-SNE permet l'affichage en 2D des images sur un répère ayant deux composantes principales en guise d'axes. Son principe repose sur la création d'une distribution de probabilité qui représente les similarités entre des données proches dans un espace de grande dimension et dans un espace de plus petite dimension.
Plus des données sont proches dans l'espace de départ, plus la probabilité qu'elles soient proches dans l'espace d'arrivée est grande.

```python
df_tsne, X_tsne = tsne_descripteurs(features_pca, photoch_acad, "label" )
```

### Génération de données

L’apprentissage par transfert repose sur le principe de profiter d’un apprentissage acquis précédemment, pour, par analogie, résoudre un problème similaire mais différent.


Les réseaux de neurones sont empilés en couches, chacune apprenant de la précédente. Il y a 4 types de couches :
- les couches de convolution, qui repèrent la présence d'un ensemble de features dans une image reçue en entrée,
- les couches de pooling qui réduisent la taille des images en conservant leurs caractéritiques importantes,
- les couches de correction ReLU, qui remplace toutes les valeurs négatives reçues en entrée par des zeros,
- les couches fully-connected qui renvoie un vecteur contenant les probabilité de l'image d'appartenir à une classe.

Les premières couches du réseau vont se spécialiser dans la reconnaissance de formes simples (lignes horizontales, lignes verticales, diagonales, coin), les suivantes vont apprendre à repérer des formes un plus complexes (rond, carré, triangle) et de fil en aiguiller, les dernières couches se verront attribuer la tâche de reconnaître le label de la photo par ses caractéristiques.
